# EDA

## Analyse temporelle

```python
df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Heures'], errors='coerce')
df = df.sort_values('datetime')

df['year'] = df['datetime'].dt.year
df['month'] = df['datetime'].dt.month
df['day'] = df['datetime'].dt.day
df['hour'] = df['datetime'].dt.hour
df['weekday'] = df['datetime'].dt.day_name(locale='fr_FR')

jours_fr = ["Lundi","Mardi","Mercredi","Jeudi","Vendredi","Samedi","Dimanche"]
df["weekday_num"] = df["datetime"].dt.dayofweek
df["weekday_fr"] = df["weekday_num"].map(dict(enumerate(jours_fr)))
df["is_weekend"] = df["weekday_fr"].isin(["Samedi","Dimanche"])
```

```python
# Évolution de la consommation dans le temps
df_daily = (
    df.dropna(subset=["Date","Consommation"])
      .groupby("Date", as_index=False)["Consommation"].mean()
)
df_daily["Date"] = pd.to_datetime(df_daily["Date"], errors="coerce")
fig1 = px.line(df_daily, x="Date", y="Consommation",
               title="Consommation moyenne quotidienne")
fig1.show()
```

```python
# Saisonnalité horaire
df_hour = (
    df.dropna(subset=["hour","Consommation"])
      .groupby("hour", as_index=False)["Consommation"].mean()
)
fig2 = px.bar(df_hour, x="hour", y="Consommation",
              title="Consommation moyenne par heure de la journée")
fig2.show()
```

```python
# Saisonnalité mensuelle
df_month = (
    df.dropna(subset=["month","Consommation"])
      .groupby("month", as_index=False)["Consommation"].mean()
)
fig3 = px.bar(df_month, x="month", y="Consommation",
              title="Consommation moyenne par mois")
fig3.show()
```

```python
# Semaine vs week-end
df_week = (
    df.dropna(subset=["is_weekend","Consommation"])
      .groupby("is_weekend", as_index=False)["Consommation"].mean()
      .replace({"is_weekend": {True: "Week-end", False: "Semaine"}})
)
fig4 = px.bar(df_week, x="is_weekend", y="Consommation",
              title="Consommation moyenne : semaine vs week-end")
fig4.show()
```

```python
# Évolution des productions principales non revouvlable (Nucléaire et gaz)
prod_cols_01 = [c for c in ["Nucléaire", "Gaz"]
             if c in df.columns]
if prod_cols_01:
    df_prod = (
        df.dropna(subset=["Date"])
          .groupby("Date", as_index=False)[prod_cols_01].mean()
    )
    df_prod["Date"] = pd.to_datetime(df_prod["Date"], errors="coerce")
    df_prod_long = df_prod.melt(id_vars="Date", value_vars=prod_cols_01,
                                var_name="source", value_name="production_MW")
    fig5 = px.line(df_prod_long, x="Date", y="production_MW", color="source",
                   title="Évolution moyenne journalière des productions par source (MW)")
    fig5.show()
```

```python
# Évolution des productions principales revouvlable
prod_cols_02 = [c for c in ["Eolien", "Solaire", "Hydraulique"]
             if c in df.columns]
if prod_cols_02:
    df_prod = (
        df.dropna(subset=["Date"])
          .groupby("Date", as_index=False)[prod_cols_02].mean()
    )
    df_prod["Date"] = pd.to_datetime(df_prod["Date"], errors="coerce")
    df_prod_long = df_prod.melt(id_vars="Date", value_vars=prod_cols_02,
                                var_name="source", value_name="production_MW")
    fig6 = px.line(df_prod_long, x="Date", y="production_MW", color="source",
                   title="Évolution moyenne journalière des productions par source")
    fig6.show()
```

```python
# Intensité carbone dans le temps
if "Taux de Co2" in df.columns:
    fig7 = px.line(df.dropna(subset=["datetime","Taux de Co2"]),
                   x="datetime", y="Taux de Co2",
                   title="Évolution du taux de CO₂ (CO2 en g/kWh)")
    fig7.show()
```

```python
# moyenne par jour de semaine et par heure (heatmap)
pivot = (df.dropna(subset=["weekday_fr","hour","Consommation"])
           .groupby(["weekday_fr","hour"], as_index=False)["Consommation"].mean())
pivot["weekday_fr"] = pd.Categorical(pivot["weekday_fr"], categories=jours_fr, ordered=True)
pivot = pivot.sort_values(["weekday_fr","hour"])

fig8 = px.density_heatmap(pivot, x="hour", y="weekday_fr", z="Consommation",
                          nbinsx=24, histfunc="avg",
                          title="Consommation moyenne par heure et jour de semaine")
fig8.show()
```

## Analyse énergétique

```python
# Composition du mix énergétique global
prod_cols = [c for c in [
    "Nucléaire", "Gaz", "Charbon", "Fioul",
    "Hydraulique", "Eolien", "Solaire", "Bioénergies"
] if c in df.columns]

df["production_totale"] = df[prod_cols].sum(axis=1)

mix_global = (
    df[prod_cols]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
    .rename(columns={"index": "source", 0: "production_moyenne"})
)

fig1 = px.pie(
    mix_global,
    names="source",
    values="production_moyenne",
    title="Répartition moyenne du mix énergétique",
    hole=0.4
)
fig1.update_traces(textposition="inside", textinfo="percent+label")
fig1.show()
```

```python
# Part des renouvelables vs fossiles
enr = ["Eolien", "Solaire", "Hydraulique", "Bioénergies"]
fossiles = ["Gaz", "Fioul", "Charbon"]

df["prod_renouvelable"] = df[enr].sum(axis=1)
df["prod_fossile"] = df[fossiles].sum(axis=1)
df["part_renouvelable"] = df["prod_renouvelable"] / df["production_totale"] * 100
df["part_fossile"] = df["prod_fossile"] / df["production_totale"] * 100

fig2 = px.line(
    df.sort_values("datetime"),
    x="datetime",
    y="part_renouvelable",
    title="Évolution de la part des énergies renouvelables dans le mix (%)"
)
fig4.show()
```

```python
# Moyenne de production par source (par heure)
df_hour_prod = (
    df.groupby("hour")[prod_cols]
    .mean()
    .reset_index()
    .melt(id_vars="hour", var_name="source", value_name="production_moyenne")
)

fig6 = px.line(
    df_hour_prod,
    x="hour",
    y="production_moyenne",
    color="source",
    title="Profil moyen horaire de production par source"
)
fig6.show()
```

## Émissions de CO₂

```python
# Contribution des sources aux émissions de CO₂
if "Taux de Co2" in df.columns:
    corr_co2 = df[prod_cols + ["Taux de Co2"]].corr()["Taux de Co2"].sort_values(ascending=False)
    corr_co2 = corr_co2.reset_index().rename(columns={"index": "source", "Taux de Co2": "corr_CO2"})

    fig8 = px.bar(
        corr_co2,
        x="source",
        y="corr_CO2",
        color="corr_CO2",
        color_continuous_scale="RdBu_r",
        title="Corrélation entre les sources d’énergie et le taux de CO₂"
    )
    fig8.show()
```

```python
# Moyenne par heure et par mois
# Par heure
df_hour = df.groupby("hour", as_index=False)["Taux de Co2"].mean()
fig3 = px.bar(df_hour, x="hour", y="Taux de Co2",
              title="Taux moyen de CO₂ par heure de la journée (gCO₂/kWh)",
              color_discrete_sequence=["indianred"])
fig3.show()
```

```python
# Par mois
df_month = df.groupby("month", as_index=False)["Taux de Co2"].mean()
fig4 = px.bar(df_month, x="month", y="Taux de Co2",
              title="Taux moyen de CO₂ par mois (gCO₂/kWh)",
              color_discrete_sequence=["indianred"])
fig4.show()
```

```python
# Heatmap CO₂ par jour de semaine et heure
jours_fr = ["Lundi","Mardi","Mercredi","Jeudi","Vendredi","Samedi","Dimanche"]
df["weekday_fr"] = pd.Categorical(df["weekday_fr"], categories=jours_fr, ordered=True)

pivot = (
    df.dropna(subset=["weekday_fr","hour","Taux de Co2"])
      .groupby(["weekday_fr","hour"], as_index=False)["Taux de Co2"].mean()
)

fig7 = px.density_heatmap(
    pivot,
    x="hour",
    y="weekday_fr",
    z="Taux de Co2",
    nbinsx=24,
    histfunc="avg",
    color_continuous_scale="RdYlGn_r",
    title="Intensité carbone moyenne par heure et jour de semaine"
)
fig7.update_layout(xaxis_title="Heure", yaxis_title="Jour de la semaine")
fig7.show()
```

## Analyse des échanges internationaux

```python
ech_cols = [c for c in df.columns if "ech." in c.lower()]
if not ech_cols:
    raise ValueError("Aucune colonne d'échanges détectée.")

# 2) Copie + conversion stricte en numérique
df_ech = df.copy()
df_ech[ech_cols] = df_ech[ech_cols].apply(pd.to_numeric, errors="coerce")

rename_map = {
    "ech._physiques": "Total (physiques)",
    "ech._comm._angleterre": "Angleterre",
    "ech._comm._espagne": "Espagne",
    "ech._comm._italie": "Ita   lie",
    "ech._comm._suisse": "Suisse",
    "ech._comm._allemagne-belgique": "Allemagne-Belgique",
}
cols_pretty = [rename_map.get(c, c) for c in ech_cols]
```

```python
ech_moy = (
    df_ech[ech_cols]
    .mean(numeric_only=True)
    .sort_values(ascending=False)
    .reset_index()
    .rename(columns={"index": "pays", 0: "ech_moyen"})
)
ech_moy["pays"] = ech_moy["pays"].map(rename_map).fillna(ech_moy["pays"])

fig1 = px.bar(
    ech_moy, x="pays", y="ech_moyen",
    color="ech_moyen", color_continuous_scale="Blues",
    title="Volume moyen d’échanges physiques par pays (MW)"
)
fig1.show()
```

```python
# Évolution temporelle des échanges physiques totaux
if "Ech. physiques" in df.columns:
    df_ech = (
        df.groupby("Date", as_index=False)["Ech. physiques"]
        .mean()
        .dropna()
    )
    df_ech["Date"] = pd.to_datetime(df_ech["Date"], errors="coerce")
    fig2 = px.line(
        df_ech,
        x="Date",
        y="Ech. physiques",
        title="Évolution des échanges physiques totaux",
        color_discrete_sequence=["royalblue"]
    )
    fig2.update_layout(yaxis_title="MW (positif = export, négatif = import)")
    fig2.show()
```
