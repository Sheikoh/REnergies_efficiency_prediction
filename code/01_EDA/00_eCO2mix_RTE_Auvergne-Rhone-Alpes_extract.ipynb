```python
from pathlib import Path
from zipfile import ZipFile
from tqdm import tqdm
import pandas as pd
```

```python
# Extraction des datasets
data_dir = Path("data/prod/zip/regional")
out_dir = Path("data/prod/unzipped/regional")
out_dir.mkdir(parents=True, exist_ok=True)

# Liste de toutes les archives zip
zip_files = sorted(data_dir.glob("*.zip"))

# Barre de progression
for z in tqdm(zip_files, desc="Extraction des fichiers", unit="archive"):
    with ZipFile(z) as zf:
        zf.extractall(out_dir)
```

```python
data_dir = Path("data/prod/unzipped/regional")

# Liste de tous les fichiers .xlsx
xlsx_files = sorted(data_dir.glob("*.xlsx"))

# Chargement et concaténation
dfs = []
for file in xlsx_files:
    df = pd.read_excel(file)
    #df["source_file"] = file.stem
    dfs.append(df)

# Concaténation de tous les DataFrames
df_all = pd.concat(dfs, ignore_index=True)

print(f"{len(dfs)} fichiers chargés")
print(df_all.shape)
df_all.head()
```

```python
df_all = df_all[~df_all["Heures"].astype(str).str.endswith(("15:00", "45:00"))]
```

```python
df_all.to_csv("data/eCO2mix_RTE_Auvergne-Rhone-Alpes_extract.csv", index=False)
```
